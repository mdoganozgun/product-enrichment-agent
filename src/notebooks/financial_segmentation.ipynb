```python
import pandas as pd
from sklearn.cluster import KMeans

# Veri yükleme
df = pd.read_csv("../../data/customer_financial_reports.csv")

# CustomerID gibi sayısal olmayan kolonları çıkar
X = df.drop(columns=["CustomerID"], errors="ignore")

# NaN değerleri düşür (sen zaten dropna yapmışsın)
X = X.dropna()

# KMeans modelini kur
cluster = KMeans(n_clusters=3, random_state=42)

# Fit + Predict
labels = cluster.fit_predict(X)

# Etiketleri DF'e ekleyelim
df_clean = df.loc[X.index].copy()  # NaN atılan satırlarla senkron tut
df_clean["Cluster"] = labels

print(df_clean.head())
```
